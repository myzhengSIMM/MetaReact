```python
from rdkit import Chem

smi = 'CC(C)(O)c1ccccc1CC[C@@H](SCC1(CC(=O)O)CC1)c1cccc(/C=C/c2ccc3ccc(Cl)cc3n2)c1'


def generate_kcf_from_smiles(mol):
    kcf_content = []

    # 添加分子名称
    kcf_content.append('COMPOUND    MoleculeName')

    # 获取原子和键的信息
    atom_count = mol.GetNumAtoms()
    kcf_content.append(f'ATOM    {atom_count}')
    
    # 原子信息
    for idx, atom in enumerate(mol.GetAtoms()):
        element = atom.GetSymbol()
        x, y, z = 0.0, 0.0, 0.0  # 默认坐标，实际中应计算3D坐标
        kcf_content.append(f'{idx+1:4d} {element:2} {x:8.3f} {y:8.3f} {z:8.3f}')
    
    # 键信息
    bond_count = mol.GetNumBonds()
    kcf_content.append(f'BOND    {bond_count}')
    
    for bond in mol.GetBonds():
        atom1 = bond.GetBeginAtomIdx() + 1
        atom2 = bond.GetEndAtomIdx() + 1
        bond_type = bond.GetBondTypeAsDouble()  # 1.0 for single, 2.0 for double, etc.
        kcf_content.append(f'{atom1:4d} {atom2:4d} {int(bond_type)}')

    # 返回构建的 KCF 内容
    return '\n'.join(kcf_content)

# 生成 KCF 内容
kcf = generate_kcf_from_smiles(mol)

# 打印 KCF 内容
print(kcf)

with open('molecule.kcf', 'w') as f:
    f.write(kcf)
```
