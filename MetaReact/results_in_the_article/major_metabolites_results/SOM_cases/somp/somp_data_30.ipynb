```python
import pandas as pd
import os
from rdkit import Chem
```

```python
def canonicalsmiles(smi):
    mol = Chem.MolFromSmiles(smi)
    smi_new = Chem.MolToSmiles(mol)
    return smi_new
```

```python
df_substrates = pd.read_csv('/home/datahouse1/raojingxin/myprojects/enzyme/metabolism/main_cases/pred_results_score_new.csv')
substrates = df_substrates['substrate'].to_list()
dict_idx2sub = dict(zip(list(range(len(substrates))),substrates))
dict_idx2sub
```

```python
dirpath = '/home/datahouse1/raojingxin/myprojects/enzyme/metabolism/SOM_cases/somp/30_main_cases'
filepathlist = os.listdir(dirpath)
substrates_new = []
enzymes_new = []
dicts_new = []

for filepath in filepathlist:
    idx = int(filepath.split('_')[0])
    substrates_new.append(dict_idx2sub[idx])
    enzyme = filepath.split('_')[1].split('.')[0]
    enzymes_new.append(enzyme)
    filepathfull = dirpath + '/' + filepath
    df = pd.read_csv(filepathfull)
    atomnums = df['Atom number'].to_list()
    ranks = df['Rang'].to_list()
    dict_atomidx2rank = dict(zip(atomnums,ranks))
    dict_atomidx2rank = {k: dict_atomidx2rank[k] for k in sorted(dict_atomidx2rank)}
    dict_atomidx2rank = {k - 1: v for k, v in dict_atomidx2rank.items()}
    dicts_new.append(dict_atomidx2rank)
substrates_new = list(map(canonicalsmiles,substrates_new))


df_new = pd.DataFrame()
df_new['substrates'] = substrates_new
df_new['enzyme'] = enzymes_new
df_new['scores'] = dicts_new
df_new
```

```python
df_new.to_pickle('/home/datahouse1/raojingxin/myprojects/enzyme/metabolism/SOM_cases/somp/somp_results_30.pickle')
```

```python
import pandas as pd

def group_by_column_to_dict(df, col_A, col_B):
    # 使用 groupby 按 col_A 分组，并将每组的 col_B 合并为一个列表
    result = df.groupby(col_A)[col_B].apply(list).to_dict()
    return result
def sum_ranks(dicts):
    # 获取所有的原子序号
    all_keys = set(key for d in dicts for key in d)
    
    # 创建一个新的字典，将原子序号的值加在一起
    result = {key: sum(d.get(key, 0) for d in dicts) for key in all_keys}
    
    return result

# 示例
result = group_by_column_to_dict(df, 'substrates', 'scores')
result_new = {}
for i,j in result.items():
    j = sum_ranks(j)
    result_new[i] = j

print(result_new)
```
