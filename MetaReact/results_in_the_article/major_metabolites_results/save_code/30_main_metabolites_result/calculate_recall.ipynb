```python
import pandas as pd
```

```python
df = pd.read_csv('results_mainproduct_top5_similaritybased_new30.csv')

def calculate_recall(true_rank_list):
    true_rank_list_top1 = [i for i in true_rank_list if i == 1]
    true_rank_list_top3 = [i for i in true_rank_list if 0 < i <= 3]
    true_rank_list_top5 = [i for i in true_rank_list if 0 < i <= 5]
    top1_recall = len(true_rank_list_top1)/len(true_rank_list)
    top3_recall = len(true_rank_list_top3)/len(true_rank_list)
    top5_recall = len(true_rank_list_top5)/len(true_rank_list)
    return top1_recall,top3_recall,top5_recall

true_rank_sygma = df['true_rank_sygma'].to_list()
true_rank_metareactseed42 = df['true_rank_metareactseed42'].to_list()
true_rank_metareactseed8079 = df['true_rank_metareactseed8079'].to_list()
true_rank_metapredictor	 = df['true_rank_metapredictor'].to_list()
true_rank_metatrans = df['true_rank_metatrans'].to_list()
true_rank_ourmodel = df['true_rank_ourmodel'].to_list()
results_sygma = calculate_recall(true_rank_sygma)
results_metareactseed42 = calculate_recall(true_rank_metareactseed42)
results_metareactseed8079 = calculate_recall(true_rank_metareactseed8079)
results_metapredictor = calculate_recall(true_rank_metapredictor)
results_metatrans = calculate_recall(true_rank_metatrans)
results_ourmodel = calculate_recall(true_rank_ourmodel)
results_ourmodel
```
