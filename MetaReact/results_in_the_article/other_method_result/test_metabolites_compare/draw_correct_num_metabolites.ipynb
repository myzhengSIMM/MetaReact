```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 示例数据
data = {
    'Method': ['Top5', 'Top10', 'Top20'],
    'MetaReact': [105, 127, 142],
    'MetaTrans': [76, 103, 116],
    'GLORYx': [54, 93, 132],
    'SyGMa': [76, 105, 127],
    'Meta-Predictor':[98,111,123]
}

df = pd.DataFrame(data)

# 将数据从宽格式转换为长格式
df_melted = pd.melt(df, id_vars=['Method'], value_vars=['MetaReact','Meta-Predictor', 'MetaTrans', 'GLORYx', 'SyGMa'],
                    var_name='Technique', value_name='Predicted_Metabolites')

# 设置Seaborn的样式
sns.set(style="whitegrid")

# 创建一个图形
plt.figure(figsize=(12, 8))

# 使用经典的学术配色方案
palette = sns.color_palette(["#66c2a5", "#fc8d62", "#8da0cb", "#e78ac3","#ffff99"])  # 自定义配色方案

# 绘制多组条形图
ax = sns.barplot(x='Method', y='Predicted_Metabolites', hue='Technique', data=df_melted, palette=palette)

# 添加标题和标签
# ax.set_title('Metabolite Identification for 65 Substrates (179 Total Metabolites)', fontsize=16)
ax.set_xlabel('')  # 隐藏横轴标签
ax.set_ylabel('Number of Identified Metabolites', fontsize=16)
plt.xticks(fontsize=14) 
# 修改图例，去掉"Technique"名字
handles, labels = ax.get_legend_handles_labels()
plt.legend(title="Total Real Metabolites: 179")
ax.legend(handles=handles, labels=labels, fontsize=14.5,title='')
plt.savefig('65_179total.png', format='png', dpi=300, bbox_inches='tight')
# 显示图表
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# 示例数据
data = {
    'Method': ['TOP5', 'TOP10', 'TOP20'],
    'ReactTrans': [104, 127, 143],
    'MetaTrans': [76, 103, 116],
    'GLORYx_1': [54, 93, 132],
    'GLORYx_2': [60, 100, 140],  # 第三组多一个子数据
    'SyGMa': [76, 105, 127]
}

df = pd.DataFrame(data)

# 将数据从宽格式转换为长格式
df_melted = pd.melt(df, id_vars=['Method'], value_vars=['ReactTrans', 'MetaTrans', 'GLORYx_1', 'GLORYx_2', 'SyGMa'],
                    var_name='Technique', value_name='Predicted_Metabolites')

# 设置Seaborn的样式
sns.set(style="whitegrid")

# 创建一个图形
plt.figure(figsize=(14, 8))

# 使用经典的学术配色方案
palette = sns.color_palette(["#D87659", "#E9C46A", "#299D8F", "#6C5B7B", "#FF5733"])  # 自定义配色方案

# 绘制多组条形图
ax = sns.barplot(x='Method', y='Predicted_Metabolites', hue='Technique', data=df_melted, palette=palette)

# 添加标题和标签
ax.set_title('Number of Identified Metabolites by Different Methods', fontsize=16)
ax.set_xlabel('')  # 隐藏横轴标签
ax.set_ylabel('Number of Identified Metabolites', fontsize=14)

# 修改图例，去掉"Technique"名字
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles=handles, labels=[label.replace('_1', '').replace('_2', '') for label in labels], title='')

# 显示图表
plt.show()
```
