1. 将predict_site.csv中的底物与ec对应起来

```python
import pandas as pd
from rdkit import Chem
from tqdm import tqdm
```

```python
def canonicalsmiles(smi):
    mol = Chem.MolFromSmiles(smi)
    smi_new = Chem.MolToSmiles(mol)
    return smi_new
```

```python
df = pd.read_csv('predict_site.csv')
substrates = df['SMILES'].to_list()
substrates = list(map(canonicalsmiles,substrates))

df_sub2ec = pd.read_csv('testMolecules_info.csv')
smiles = df_sub2ec['SMILES'].to_list()
smiles = list(map(canonicalsmiles,smiles))
ecs = df_sub2ec['Enzyme'].to_list()
dict_smiles2ecs = dict(zip(smiles,ecs))

ecs_new = [dict_smiles2ecs[i] for i in substrates]
df['SMILES'] = substrates
df['ecs'] = ecs_new
df
```

```python
df.to_csv('predict_site_withec.csv',index=None)
```

2. 查看rdkit原子和键序号

```python
from rdkit import Chem
from rdkit.Chem import Draw

# 输入 SMILES 字符串
smiles = "CCO"  # 这是一个乙醇的示例，请替换为你的 SMILES

# 从 SMILES 创建分子对象
mol = Chem.MolFromSmiles(smiles)

# 标注原子序号
for atom in mol.GetAtoms():
    atom.SetProp('atomNote', str(atom.GetIdx()))

# 标注键序号
for bond in mol.GetBonds():
    bond_idx = bond.GetIdx()
    bond.SetProp('bondNote', str(bond_idx))

# 绘制分子并保存为图像
img = Draw.MolToImage(mol, size=(400, 400), kekulize=True)
img.show()  # 显示图像
```
