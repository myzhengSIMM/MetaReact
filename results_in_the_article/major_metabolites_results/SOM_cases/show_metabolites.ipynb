1. 查看SOM数据集的结构

```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import Draw

df = pd.read_csv('predict_site.csv')
substrates = df['SMILES'].to_list()
metabolites = df['metabolite'].to_list()
preds = df['predict_site'].to_list()
def mols_2_picture(idx,col='substrates'):
    if col == 'substrates':
        smi = substrates[idx]
        smi = [smi]
        mol_list = [Chem.MolFromSmiles(smiles) for smiles in smi]
        img = Draw.MolsToGridImage(mol_list, molsPerRow=1, subImgSize=(500, 500))
    if col == 'metabolites':
        metabolite = metabolites[idx]
        metabolitelist = metabolite.split('|')
        metabolitelist = list(set(metabolitelist))
        smiles_list = metabolitelist
        mol_list = [Chem.MolFromSmiles(smiles) for smiles in smi]
        img = Draw.MolsToGridImage(mol_list, molsPerRow=1, subImgSize=(500, 500))        
    if col == 'preds':
        pred = preds[idx]
        predlist = pred.split('|')
        predlist = list(set(predlist))
        smiles_list = predlist
        mol_list = [Chem.MolFromSmiles(smiles) for smiles in smiles_list]
        img = Draw.MolsToGridImage(mol_list, molsPerRow=2, subImgSize=(500, 500))
    return img
```

```python
n = 1
```

```python
mols_2_picture(n)
```

```python
mols_2_picture(n,col='metabolites')
```

```python
mols_2_picture(n,col='preds')
```
