```python
import json
import os
import pandas as pd

dirpath1 = '/home/datahouse1/raojingxin/myprojects/enzyme/metabolism/SOM_cases/xenosite'
pathlist1 = os.listdir(dirpath1)
pathlist1 = [dirpath1+'/'+i for i in pathlist1 if i.endswith('.txt')]

dirpath2 = '/home/datahouse1/raojingxin/myprojects/enzyme/metabolism/SOM_cases/xenosite/add_unnamed_files'
pathlist2 = os.listdir(dirpath2)
pathlist2 = [dirpath2+'/'+i for i in pathlist2 if i.endswith('.txt')]

pathlist = pathlist1 + pathlist2

smiles_new = []
models_new = []
atomscores_new = []
bondscores_new = []

# pathlist = ['unnamed (1).txt']

for path in pathlist:

    try:
        # 读取文件内容
        with open(path, 'r') as file:
            data = json.load(file)

        # 提取 SMILES
        smiles = data.get('smiles')

        # 提取每个 model 的信息
        models_info = []

        for result in data['results']:
            model = result.get('model')
            atom = result.get('atom')
            bonds = result.get('bond')

            # 保存每个 model 的详细信息
            smiles_new.append(smiles)
            models_new.append(model)
            atomscores_new.append(atom)
            bondscores_new.append(bonds)
    except:
        print(path)
    

# 打印结果

# print("SMILES:", smiles)
# for model_info in models_info:
#     print("\nModel:", model_info['model'])
#     print("Atoms:", model_info['atom'])
#     print("Bonds:", model_info['bond'])

df_new = pd.DataFrame()
df_new['substrates'] = smiles_new
df_new['enzyme'] = models_new
df_new['atomscores'] = atomscores_new
df_new['bondscores'] = bondscores_new
df_new
```

```python
from rdkit import Chem
def canonicalsmiles(smiles):
    mol = Chem.MolFromSmiles(smiles)
    smi_new = Chem.MolToSmiles(mol, isomericSmiles=False)
    return smi_new
```

```python
# 输出的结果会失去立体构象
smiles_now = list(set(df_new['substrates'].to_list()))
smiles_now = list(map(canonicalsmiles,smiles_now))

df = pd.read_csv('/home/datahouse1/raojingxin/myprojects/enzyme/metabolism/SOM_cases/predict_site_withec.csv')
smiles_shouldbe = df['SMILES'].to_list()
smiles_shouldbe = list(map(canonicalsmiles,smiles_shouldbe))

smiles_out = [i for i in smiles_shouldbe if i not in smiles_now]
for i in smiles_out:
    print(i)
```

```python
df =  pd.read_pickle('xenosite_results.pickle')
atomscores = df['atomscores'].to_list()

# 使用字典推导式生成字典
atomscores = [{idx: score for idx, score in enumerate(i)} for i in atomscores]
df['atomscores'] = atomscores

def group_by_column_to_dict(df, col_A, col_B):
    # 使用 groupby 按 col_A 分组，并将每组的 col_B 合并为一个列表
    result = df.groupby(col_A)[col_B].apply(list).to_dict()
    return result
def sum_ranks(dicts):
    # 获取所有的原子序号
    all_keys = set(key for d in dicts for key in d)
    
    # 创建一个新的字典，将原子序号的值加在一起
    result = {key: sum(d.get(key, 0) for d in dicts) for key in all_keys}
    
    return result
def assign_ranks(atom_scores):
    # 按分数降序排序，并获取排序后的原子序号
    sorted_atoms = sorted(atom_scores.items(), key=lambda x: x[1], reverse=True)
    
    # 创建一个字典，保存原子序号对应的排名
    rank_dict = {}
    for rank, (atom, _) in enumerate(sorted_atoms):
        rank_dict[atom] = rank + 1
    
    # 保持原子序号从小到大的顺序，并更新为对应的排名
    ranked_atoms = {atom: rank_dict[atom] for atom in sorted(atom_scores)}
    
    return ranked_atoms

atomscores = df['atomscores'].to_list()
atomscores = list(map(assign_ranks,atomscores))
df['atomscores'] = atomscores
# df.to_pickle('xenosite_results_withrank.pickle')
# 示例
result = group_by_column_to_dict(df, 'substrates', 'atomscores')
result_new = {}
for i,j in result.items():
    j = sum_ranks(j)
    result_new[i] = j

print(result_new)
```
