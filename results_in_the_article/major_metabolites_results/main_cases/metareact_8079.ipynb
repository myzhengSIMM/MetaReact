```python
from rdkit import Chem
def canonicalsmiles(smi):
    mol = Chem.MolFromSmiles(smi)
    smi_new = Chem.MolToSmiles(mol)
    return smi_new
```

```python
import pandas as pd
df = pd.read_pickle('pred_results_substrate_truemains.pickle')
true_mains = df['true_mains'].to_list()
substrate_old = df['substrate'].to_list()
substrate_old = list(map(canonicalsmiles,substrate_old))
dict_sub2true = dict(zip(substrate_old,true_mains))
dict_sub2true
```

```python
df_metareact_8079 = pd.read_csv('MetaReact_seed8079.csv')
predict = df_metareact_8079['predict'].to_list()
substrate = df_metareact_8079['substrate'].to_list()
substrate = list(map(canonicalsmiles,substrate))
predict = [i.split('|') for i in predict]
dict_sub2predict = dict(zip(substrate,predict))
dict_sub2predict
```

```python
df['predict_8079'] = [dict_sub2predict[i] for i in substrate_old]
df
```

```python
import pandas as pd
from rdkit import Chem
from rdkit.DataStructs import TanimotoSimilarity
from rdkit.Chem import AllChem

def get_similarity(smiles1, smiles2):
    mol1 = Chem.MolFromSmiles(smiles1)
    mol2 = Chem.MolFromSmiles(smiles2)
    if mol1 and mol2:
        fp1 = AllChem.GetMorganFingerprintAsBitVect(mol1, 2, nBits=2048)
        fp2 = AllChem.GetMorganFingerprintAsBitVect(mol2, 2, nBits=2048)
        return TanimotoSimilarity(fp1, fp2)
    return 0
```

```python
true_mains = df['true_mains'].to_list()
predict_8079 = df['predict_8079'].to_list()
top_num = 5
correct_num = 0
dict_true2top = {}
for i in range(len(predict_8079)):
    true_main = true_mains[i]
    true_main = '|'.join(true_main)

    predict_8079_i = predict_8079[i]
    predict_8079_i_topn = predict_8079_i[:top_num]
    predict_8079_i_topn = list(map(canonicalsmiles,predict_8079_i_topn))

    for idx,j in enumerate(predict_8079_i_topn):
        if get_similarity(true_main,j) == 1:
            correct_num = correct_num + 1
            dict_true2top[true_main] = idx + 1
            print(f'{i}: top{idx}')
            break
correct_num/len(predict_8079)
```

```python
true_mains = ['|'.join(i) for i in true_mains]
df['rank'] = [dict_true2top.get(i,0) for i in true_mains]
df
```
